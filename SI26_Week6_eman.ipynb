```python
# ============================================================
# PROJECT 2 — FINAL LABELING + VALIDATION + CSV
# ============================================================

import pandas as pd
import re

# ------------------------------------------------------------
# 1. Combine all 157 sentences
# ------------------------------------------------------------

all_sentences = (
    data
    + [{'sentence': s} for s in raw_sentences]
    + [{'sentence': s} for s in additional_sentences]
)

print("Total sentences:", len(all_sentences))


# ------------------------------------------------------------
# 2. Create word lists for sentences without them
# ------------------------------------------------------------

for item in all_sentences:
    if 'words' not in item:
        item['words'] = item['sentence'].split()


# ------------------------------------------------------------
# 3. English words
# ------------------------------------------------------------

english_words = set("""
i you he she it we they me my your his her our their
this that these those a an the
is am are was were be been being
have has had do does did
can could will would should may might must
want need know think feel like love hate
make made get got go going gone come coming
call calling send check checking watch watching
talk talking help use used using
start starting finish finished complete completed
work working study studying learn learning
try trying wait waiting
look looks looking stay staying
keep kept take taking give giving
tell told say said run running
happen happened happening
and or but if then so because for from to
with of in on at by about until since into through
after before while than
not don't doesn't didn't can't won't isn't
just still already now later again also
very really literally seriously honestly actually
almost more most much many enough only
right wrong back away outside
what when where why how who which
good bad great easy difficult busy tired ready
important boring emotional expensive healthy clean
final proper positive negative real solid simple
big small free late early worried focused worth okay
project presentation assignment university college school
exam class lecture internship deadline meeting meetings
task interface website app data roadmap solution problem
idea course student students id fyp srs ppt
phone laptop internet file download upload
reply video show movie game editing search
google apple ai sheets suggest
dinner lunch break diet traffic weather shopping
plan plans finalize tag attitude change job
time ending form season
people paper gains king squad
opposition treasury benches government accountant
private hospitals party rich affiliated
democracy contribute launch condemnations exams types taken
once done said thank god
anything everything nothing something
somewhere anywhere kind thing things
teaching lot bored
middle interview crying getting kinda
message important version
sports play rules dare
""".lower().split())


# ------------------------------------------------------------
# 4. Roman Urdu words
# ------------------------------------------------------------

urdu_words = set("""
aaj kal mujhe mera meri mere main mein me
tum tumhe tumhara tumhari tujhe tera teri
yaar bhai baji mummy ustaad beta
kya kyun kyo ka ke ki ko se pe par
hai hain ho hoga hona hone tha thi the hoon hun
ab abhi phir bhi nahi nhy nhi na toh to aur bas
kuch koi sab sari sara ye yeh wo woh
iska uska unka usne mene maine ne
apke apna apni apny aap
ghar ghr din raat subah sham
bohot bahut thora thoda zyada
acha achhi jagah baat
kar karo karte karna karni karungi
krna krti krte krne krungi
gaya gayi gye gya gaye
jata jati jao jaana jana jaye
raha rahi rha rhi rai
lag lagta lagti lagrha
samajh smj mujhe mjy muje
pata pucho puchna chahiye
liye
khana kha khali pi pehle
neend aayi aaya aata aate
waala wala walo valo
baki baaki aik ek dosre
darmeyaan kisi
mila milne milta
waja wajah samne sath
shuru suru chup reh
mazak dil dukh kaan
zaya hathiyar sawaal
zimmedari
derhe pta krwarhe akele
mama agai bethi
auntio aye
bnai bnwalena bna
batai bnnega jawab deta
khul khula khulna
mne emre glti
tehgya bje tb
usmai khuch mushkil nai hy
agr krr k b yaa
aplogo jotmhare tmhare stah
ais sai smjhugi
bass thodi
waise
inka sale log
chaat raha
kari
sarkar
""".lower().split())


# ------------------------------------------------------------
# 5. Label ONLY the 103 unlabeled sentences
# ------------------------------------------------------------

for item in all_sentences:

    # Your original 54 labels stay untouched
    if 'labels' in item:
        continue

    labels = []

    for word in item['words']:

        # Remove punctuation
        clean = re.sub(r"^[^\w']+|[^\w']+$", "", word.lower())

        # Number
        if clean.isdigit():
            labels.append("URD")

        # Clearly English
        elif clean in english_words:
            labels.append("ENG")

        # Clearly Roman Urdu
        elif clean in urdu_words:
            labels.append("URD")

        # Common English contractions
        elif clean in {"i'll", "we'll", "you'll", "it's", "that's",
                       "i'm", "you're", "they're", "we're", "he's",
                       "she's", "wasn't", "weren't", "don't", "can't"}:
            labels.append("ENG")

        # Unknown word:
        # default to URD because the dataset is Roman Urdu code-switching
        else:
            labels.append("URD")

    item['labels'] = labels


# ------------------------------------------------------------
# 6. FINAL VALIDATION
# ------------------------------------------------------------

print("\n========== VALIDATION ==========")

errors = []

for i, item in enumerate(all_sentences):

    if len(item['words']) != len(item['labels']):
        errors.append(i)

print("Total sentences:", len(all_sentences))
print("Sentences with labels:", sum('labels' in x for x in all_sentences))

if len(errors) == 0:
    print("✅ ALL WORD/LABEL COUNTS MATCH")
else:
    print("❌ MISMATCHES FOUND:", len(errors))

    for i in errors:
        print("\nSentence:", all_sentences[i]['sentence'])
        print("Words:", len(all_sentences[i]['words']))
        print("Labels:", len(all_sentences[i]['labels']))


# ------------------------------------------------------------
# 7. Create final flat CSV
# ------------------------------------------------------------

rows = []

for item in all_sentences:

    for word, label in zip(item['words'], item['labels']):

        rows.append({
            'sentence': item['sentence'],
            'word': word,
            'label': label
        })

df = pd.DataFrame(rows)

df.to_csv(
    'dataset.csv',
    index=False,
    encoding='utf-8'
)


# ------------------------------------------------------------
# 8. Final statistics
# ------------------------------------------------------------

print("\n========== FINAL DATASET ==========")

print("Unique sentences:", df['sentence'].nunique())
print("Total word entries:", len(df))

print("\nLabel distribution:")
print(df['label'].value_counts())

print("\n✅ dataset.csv CREATED SUCCESSFULLY!")
```
